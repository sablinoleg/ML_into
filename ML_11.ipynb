 1. Загрузите данные close_prices.csv. В этом файле приведены цены
 акций 30 компаний на закрытии торгов за каждый день периода.

```python
import pandas as pd

df = pd.read_csv('close_prices.csv', parse_dates=['date'], index_col='date')

djia = pd.read_csv('djia_index.csv', parse_dates=['date'], index_col='date')
```

 2. На загруженных данных обучите преобразование PCA с числом
 компоненты равным 10. Скольких компонент хватит, чтобы объяснить 90% дисперсии?

```python
from sklearn.decomposition import PCA

# для 10 компонетов
pca = PCA(n_components=10)
pca.fit(df)

# объясняет более 90% дисперсии
import numpy as np
explained_variance = np.cumsum(pca.explained_variance_ratio_)
num_components_90 = np.argmax(explained_variance >= 0.9) + 1
print(num_components_90)
```

 3. Применитепостроенное преобразование к исходным данным и возь
мите значения первой компоненты.

```python
X_pca = pca.transform(df)
first_component = X_pca[:, 0]
```

 4. Загрузите информацию об индексе Доу-Джонса из файла djia_prices.csv.
 Чему равна корреляция Пирсона между первой компонентой и индексом Доу-Джонса?

```python
correlation = np.corrcoef(first_component, djia['^DJI'])[0, 1]
print(round(correlation, 2))
```

 5. Какая компания имеет наибольший вес в первой компоненте?

```python
weights = pca.components_[0]

company_weights = pd.Series(weights, index=df.columns)

most_influential = company_weights.abs().idxmax()
print(most_influential)
```
